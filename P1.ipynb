```python
#1
import os

# Define the path to the directory containing the text files
corpus_root = './US_Inaugural_Addresses'

# List to store the content of each document
documents = []

# Iterate over every file in the directory
for filename in os.listdir(corpus_root):
    # Check if the file is a text file
    if filename.endswith('.txt'):
        # Create the full filepath by joining the directory path and filename
        file_path = os.path.join(corpus_root, filename)
        
        # Open the file for reading with the correct encoding
        with open(file_path, 'r', encoding='windows-1252') as file:
            # Read the content of the file
            doc = file.read()
            
            # Convert the text to lowercase
            doc = doc.lower()
            
            # Append the processed text to the documents list
            documents.append(doc)

# At this point, 'documents' contains all the text files' contents in lowercase
print("Number of documents read:", len(documents))
```

```python
#2
from nltk.tokenize import RegexpTokenizer

# Initialize the tokenizer with a regular expression to capture words
tokenizer = RegexpTokenizer(r'[a-zA-Z]+')

# Assuming 'documents' is the list containing all your documents
# which you have previously read and converted to lowercase
tokenized_documents = []

for doc in documents:
    # Tokenize each document
    tokens = tokenizer.tokenize(doc)
    # Append the tokenized document to the list
    tokenized_documents.append(tokens)

# If you want to see the tokens of the first document as an example
print("Tokens of the first document:", tokenized_documents[0])
```

```python
#3
import nltk
from nltk.corpus import stopwords

# Download the stopwords corpus if it hasn't been downloaded yet
nltk.download('stopwords')

# Load the English stopwords
stop_words = set(stopwords.words('english'))

# Assuming 'tokenized_documents' is your list of lists, where each sublist contains the tokens of a document
filtered_documents = []

# Iterate through each tokenized document and filter out the stopwords
for tokens in tokenized_documents:
    filtered_tokens = [token for token in tokens if token not in stop_words]
    filtered_documents.append(filtered_tokens)

# If you want to see the tokens of the first document after stopword removal as an example
print("Tokens of the first document after stopword removal:", filtered_documents[0])
```

```python
#4
from nltk.stem.porter import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Assuming 'filtered_documents' is your list of documents with stopwords removed
stemmed_documents = []

# Stem each token in each document
for tokens in filtered_documents:
    stemmed_tokens = [stemmer.stem(token) for token in tokens]
    stemmed_documents.append(stemmed_tokens)

# If you want to see the tokens of the first document after stemming
print("Tokens of the first document after stemming:", stemmed_documents[0])
```

```python
#5
import math
from collections import Counter

# Number of documents
N = len(stemmed_documents)

# Calculate document frequency for each token across all documents
document_frequency = Counter()

for tokens in stemmed_documents:
    unique_tokens = set(tokens)
    for token in unique_tokens:
        document_frequency[token] += 1

# Function to calculate TF-IDF for a single document
def calculate_tfidf(tokens):
    tf = Counter(tokens)  # Term frequency in the document
    tfidf = {}
    doc_length = 0
    
    for token, count in tf.items():
        term_frequency = 1 + math.log10(count)
        inverse_doc_freq = math.log10(N / document_frequency[token])
        tfidf[token] = term_frequency * inverse_doc_freq
        doc_length += tfidf[token] ** 2

    # Normalize the TF-IDF vector to length 1
    doc_length = math.sqrt(doc_length)
    for token in tfidf:
        tfidf[token] /= doc_length

    return tfidf

# Calculate TF-IDF vectors for all documents
tfidf_vectors = [calculate_tfidf(tokens) for tokens in stemmed_documents]

# Optionally print the TF-IDF vector of the first document
print("TF-IDF vector for the first document:", tfidf_vectors[0])
```

```python
#6
from nltk.tokenize import RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
import math

# Initialize the tokenizer, stopword set, and stemmer
tokenizer = RegexpTokenizer(r'[a-zA-Z]+')
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

# Function to preprocess and calculate the TF-IDF vector for a query
def calculate_query_vector(query):
    # Tokenize, remove stopwords, and stem the query
    tokens = tokenizer.tokenize(query.lower())
    filtered_tokens = [token for token in tokens if token not in stop_words]
    stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]

    # Calculate term frequencies for the query
    tf_query = Counter(stemmed_tokens)
    tfidf_query = {}
    query_length = 0

    # Compute TF-IDF using the same document frequency and N from your corpus
    for token, count in tf_query.items():
        if token in document_frequency:  # Ensure the token was in the original corpus
            term_frequency = 1 + math.log10(count)
            inverse_doc_freq = math.log10(N / document_frequency[token])
            tfidf_query[token] = term_frequency * inverse_doc_freq
            query_length += tfidf_query[token] ** 2

    # Normalize the query vector
    query_length = math.sqrt(query_length)
    for token in tfidf_query:
        tfidf_query[token] /= query_length

    return tfidf_query

# Example usage
query = "example query text"
query_vector = calculate_query_vector(query)
print("Query TF-IDF Vector:", query_vector)
```

```python
#7
import math

# Function to compute cosine similarity between two vectors
def cosine_similarity(vec1, vec2):
    intersection = set(vec1.keys()) & set(vec2.keys())
    numerator = sum([vec1[x] * vec2[x] for x in intersection])

    sum1 = sum([vec1[x]**2 for x in vec1.keys()])
    sum2 = sum([vec2[x]**2 for x in vec2.keys()])
    denominator = math.sqrt(sum1) * math.sqrt(sum2)

    if not denominator:
        return 0.0
    else:
        return float(numerator) / denominator

# Assume 'tfidf_vectors' contains all your document vectors and 'query_vector' is your precomputed query TF-IDF vector
def find_most_similar_document(query_vector, document_vectors):
    max_similarity = 0
    most_similar_document_index = -1

    # Compute cosine similarity with each document
    for index, doc_vector in enumerate(document_vectors):
        similarity = cosine_similarity(query_vector, doc_vector)
        if similarity > max_similarity:
            max_similarity = similarity
            most_similar_document_index = index

    return most_similar_document_index, max_similarity

# Example usage
query = "example query text"
query_vector = calculate_query_vector(query)  # Assuming this function is already defined and used as shown earlier
most_similar_index, highest_similarity = find_most_similar_document(query_vector, tfidf_vectors)

print("The most similar document index is:", most_similar_index)
print("With a cosine similarity score of:", highest_similarity)
```

```python
#7.1
from collections import defaultdict
import operator

# Assuming tfidf_vectors is a list of dictionaries where each dictionary represents the TF-IDF vector for a document
# and document identifiers are simply the index positions in this list

# Create a dictionary to store the postings list for each token
postings_lists = defaultdict(list)

for doc_id, tfidf_vector in enumerate(tfidf_vectors):
    for token, weight in tfidf_vector.items():
        postings_lists[token].append((doc_id, weight))

# Sort each postings list in descending order by the TF-IDF weight
for token, postings in postings_lists.items():
    postings.sort(key=lambda x: x[1], reverse=True)  # Sort by weight in descending order

# Example to print the postings list for a specific token
token_to_query = 'example'  # Change this to the token you want to query
if token_to_query in postings_lists:
    print(f"Postings list for the token '{token_to_query}':", postings_lists[token_to_query])
else:
    print(f"No postings found for the token '{token_to_query}'")
```

```python
#7.2
def get_top_10_postings(query):
    # Tokenize, remove stopwords, and stem the query as before
    tokens = tokenizer.tokenize(query.lower())
    filtered_tokens = [token for token in tokens if token not in stop_words]
    stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]

    # Dictionary to store the top-10 postings for each token
    top_10_postings = {}

    for token in stemmed_tokens:
        if token in postings_lists:  # Check if the token exists in the postings lists
            # Get the top-10 elements if there are at least 10, otherwise get all
            top_10_postings[token] = postings_lists[token][:10]
        else:
            top_10_postings[token] = []  # Return an empty list if the token is not found

    return top_10_postings

# Example usage with a sample query
query = "george washington example"
top_10_results = get_top_10_postings(query)
print("Top-10 Postings for each token in the query:")
for token, postings in top_10_results.items():
    print(f"Token: {token}, Postings: {postings}")
```

```python
#7.3
def calculate_document_score(query_vector, postings_lists, doc_id):
    # Calculate the dot product between query vector and document vector for common tokens
    score = 0
    for token, q_weight in query_vector.items():
        for doc, d_weight in postings_lists[token]:
            if doc == doc_id:
                score += q_weight * d_weight
                break

    return score

# Assuming you have a function to preprocess and get the query vector, and a dictionary 'top_10_postings' from above
query = "example search query text"
query_vector = calculate_query_vector(query)
top_10_postings = get_top_10_postings(query)

# Assuming you want to calculate the score for a specific document ID that appears in the postings
doc_id_to_score = 5  # Change this to your specific document ID

# Find and calculate the cosine similarity score for the document
document_score = calculate_document_score(query_vector, top_10_postings, doc_id_to_score)
print(f"Cosine similarity score for document {doc_id_to_score} is: {document_score}")
```

```python
#7.4
def calculate_upper_bound_score(query_vector, postings_lists, doc_id):
    upper_bound_score = 0
    # Iterate through each token in the query vector
    for token, q_weight in query_vector.items():
        # Check if the token has a postings list and the document is not in the top-10
        if token in postings_lists:
            # Find the document in the postings list or use the weight of the 10th element as an upper bound
            found = False
            for doc, d_weight in postings_lists[token]:
                if doc == doc_id:
                    upper_bound_score += q_weight * d_weight
                    found = True
                    break
            if not found and len(postings_lists[token]) >= 10:
                # Use the weight of the 10th element (assuming it is sorted and the 10th element exists)
                _, weight_10th = postings_lists[token][9]  # 0-based index, 10th element is at index 9
                upper_bound_score += q_weight * weight_10th
        else:
            # If no postings list exists for this token, contribute zero to the score
            continue

    return upper_bound_score

# Example usage
query = "example search query text"
query_vector = calculate_query_vector(query)  # Assuming the function to get the query vector is already defined
top_10_postings = get_top_10_postings(query)  # Assuming the function to get top-10 postings is already defined
doc_id_to_score = 1  # Example document ID to calculate the upper bound score

# Calculate the upper bound score for the document
document_upper_bound_score = calculate_upper_bound_score(query_vector, top_10_postings, doc_id_to_score)
print(f"Upper bound cosine similarity score for document {doc_id_to_score} is: {document_upper_bound_score}")
```

```python
#7.5
def find_best_document(query_vector, top_10_postings):
    # Calculate actual scores for all documents in top-10 and upper-bound scores for others
    document_scores = {}  # Stores actual scores
    upper_bound_scores = {}  # Stores upper bound scores

    # Calculate actual scores for documents in top-10 postings
    for token, postings in top_10_postings.items():
        for doc_id, weight in postings:
            if doc_id not in document_scores:
                document_scores[doc_id] = 0
            document_scores[doc_id] += query_vector.get(token, 0) * weight

    # Identify all documents from postings lists to check upper bounds
    all_docs = set(doc for postings in top_10_postings.values() for doc, _ in postings)

    # Calculate upper-bound scores for documents not in top-10 actual scores
    for token, postings in top_10_postings.items():
        if len(postings) == 10:  # Ensure there are exactly 10 to consider the last one as an upper bound
            last_weight = postings[-1][1]
            for doc_id in all_docs - set(doc for doc, _ in postings):
                if doc_id not in upper_bound_scores:
                    upper_bound_scores[doc_id] = 0
                upper_bound_scores[doc_id] += query_vector.get(token, 0) * last_weight

    # Determine the document with the highest actual score that is better than all upper-bound scores
    best_document = None
    best_score = -1
    for doc_id, score in document_scores.items():
        if score > upper_bound_scores.get(doc_id, 0) and score > best_score:
            best_document = doc_id
            best_score = score

    return best_document, best_score

# Example usage with a sample query
query = "example search query text"
query_vector = calculate_query_vector(query)  # Assuming this function is already defined
top_10_postings = get_top_10_postings(query)  # Assuming this function to get top-10 postings is already defined

# Find the best document based on actual and upper-bound scores
best_doc, best_doc_score = find_best_document(query_vector, top_10_postings)
print(f"The best document ID is: {best_doc} with a score of: {best_doc_score}")
```
